```python
import json
for name in ['baseline', 'rmsnorm', 'swiglu', 'nope', 'rope', 'gqa2']:
    rows = [json.loads(line) for line in open(f'out/{name}/metrics.jsonl') if 'val_loss' in line]
    best = min(rows, key=lambda x: x['val_loss'])
    print(f'{name}: best val={best["val_loss"]:.4f} at step {best["iter"]}')

from IPython.display import Image, display
display(Image('results/loss_comparison.png'))
```
